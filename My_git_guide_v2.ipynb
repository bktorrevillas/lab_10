# What is git?
Git is a version control system. It is a collection of programs that tracks changes to files over time; typically these files are scripts or source code. It was created by Linus Torvalds in 2005 to manage development of the Linux operating system.

## What does git do?
* **History:** Every saved change to a file (called a “commit”) is recorded along with a note of who made the change, when, and a message explaining why. You can view any earlier commit of your code and even fully restore it.
* **Branching:** You can create a separate line of work (called a “branch”) to try a feature or fix without touching the main version, then merge it back when it’s ready.
* **Collaboration:** Many people can work on the same project at once, and Git helps combine their changes and flags conflicts when two people edit the same lines.

## Git terminology
* **Repository (repo):** The project folder plus its full history. The full history is stored in a hidden .git subdirectory.
* **Commit:** A snapshot of your files at a point in time.
* **Branch:** A movable pointer to a line of commits. The default is usually 'main' for main branch.
* **Remote:** A copy of the repo hosted elsewhere, such as on GitHub, GitLab, or Bitbucket.

## Git vs GitHub
One common point of confusion: Git is the tool, while GitHub is a website that hosts Git repositories and adds features like pull requests and issue tracking. You can use Git without GitHub.

## Basic workflow
* git init                  # start a repo
* git add file.txt          # stage a change
* git commit -m "message"   # record it
* git push                  # send commits to a remote e.g. GitHub
* git pull                  # fetch and merge commits from a remote

## Why are we learning git?
You are not going to be software developers, and you do not need to learn everything git can do (we will skip branching and merging entirely). For this class, git is mainly a tool for **sharing code**:

* **Back up your scripts** on GitHub, so that one deleted file or one lost cluster account does not cost you months of work.
* **Move code between computers** (your laptop, Kamiak, a collaborator's machine) without emailing files or running `scp`.
* **Get code from other people**, including code shared by your instructor or published with a paper, with a single command.
* **Keep a history** so that when a script that worked yesterday is broken today, you can see exactly what changed.

### How the pieces fit together
Your files move through several areas on the way to GitHub. Each git command moves changes one step along:

```
 working directory  --git add-->  staging area  --git commit-->  local repository  --git push-->  GitHub (remote)
 (your files)                     (what will be                   (your history,                   (shared copy)
                                   in the next commit)             in the .git folder)
        <------------------------------------------ git pull  <------------------------------------
```

## Untracked vs tracked files
An untracked file is any file in your project folder that git is not yet tracking. These are files you've created or copied into the folder, but haven't told git to watch.

A tracked file is a file that git is watching for changes. To make a file tracked, you need to **stage** it.

## Go back to an older version of a file
Every commit is a saved version. `git log --oneline` shows the hash of each one, and `git show` displays what a commit changed:

````bash
git show --stat HEAD
````

To bring back the version of `index.html` from an earlier commit, give `git restore` the hash with `--source`. (`HEAD~1` means "one commit before the latest one", so you do not have to look up the hash.) The old version appears in your working directory as a modification, which you can then commit, or discard with another `git restore index.html`.

````bash
git restore --source=HEAD~1 index.html
cat index.html
````

````bash
git status
````

````bash
# put the latest version back
git restore index.html
git status
````

# Ignoring files with `.gitignore`
Not everything in your project folder belongs in git. Git is designed for **small text files like scripts**, and you should leave out:

* temporary files and folders that programs create automatically (like `.ipynb_checkpoints/` from Jupyter, or `__pycache__/` from Python),
* output files such as Slurm logs (`slurm-*.out`),
* large data files (sequencing reads, genomes, BLAST databases). GitHub will refuse files over 100 MB, and your data already has a home on Kamiak, and
* **anything private**: passwords, API keys, SSH private keys, or unpublished data. Once something has been pushed to GitHub, assume it is permanently public, even if you delete it later.

A file named `.gitignore` lists patterns for files that git should pretend do not exist. Create one now (the `cat > file << 'EOF'` pattern writes everything up to the line `EOF` into a file):

````bash
cat > .gitignore << 'EOF'
# Jupyter and Python
.ipynb_checkpoints/
__pycache__/

# Slurm output and other logs
slurm-*.out
*.log
EOF
````

````bash
cat .gitignore
````

Now `git status` no longer lists `.ipynb_checkpoints/`:

````bash
git status
````

.gitignore itself is a regular file that you want to share, so add and commit it:

````bash
git add .gitignore
git commit -m "Add .gitignore"
````

### A shortcut: `git add .`
`git add .` stages **every** new and modified file in the current directory (except ignored files). It is convenient, but it will stage anything you forgot about, so look at `git status` first.

## Why SSH?
When you push code to GitHub, GitHub needs to know that you are really you. GitHub does not accept account passwords for git commands any more, so you have two options: a personal access token over HTTPS, or an **SSH key**. You already know how SSH keys work: GitHub will hold your **public** key, and your computer proves it has the matching **private** key. We will use SSH.

## Step 1. Check for an SSH key on Kamiak
Here is a detail that trips people up. `git push` runs **on Kamiak**, so the SSH private key has to be on Kamiak too. The key that you use to log in to Kamiak from your laptop is stored on your laptop, and it will not work here. See what keys exist on Kamiak:

````bash
ls -l ~/.ssh
````

Look for a pair of files with the same name, one ending in `.pub`, such as `id_ed25519` and `id_ed25519.pub`. The file without `.pub` is your **private key**, and you should never copy it, email it, post it, or commit it to git. The file ending in `.pub` is the **public key**, and it is the one you give to GitHub.

**If you do not have a key pair on Kamiak**, open a **terminal** (not this notebook, because the command below asks you to type a passphrase) and make one:

    ssh-keygen -t ed25519 -C "your.email@wsu.edu"

Press Enter to accept the default file location, and choose a passphrase (or leave it empty, but a passphrase is better practice). If your key has a passphrase, run the `git push` commands in this notebook only after unlocking the key in a terminal, because a notebook cell has no way to ask you for it. When in doubt, run those commands in a terminal instead.

## Step 2. Copy your public key
Print your public key. (If your file has a different name, such as `id_rsa.pub`, change it below.) It is a single line that starts with `ssh-ed25519`. Select the **whole line** and copy it.

````bash
cat ~/.ssh/id_ed25519.pub
````

## Step 3. Add the public key to your GitHub account
In your web browser:

1. Log in to GitHub, click your profile picture (top right), and choose **Settings**.
2. In the left sidebar, choose **SSH and GPG keys**.
3. Click **New SSH key**.
4. For **Title**, enter something that tells you which computer the key is on, e.g., `Kamiak`.
5. Leave **Key type** as *Authentication Key*.
6. Paste your public key into the **Key** box, and click **Add SSH key**.

## Step 4. Test the connection
This command asks GitHub to authenticate you using your key. `-T` means "do not try to open a shell", because GitHub does not give shell access. `-o StrictHostKeyChecking=accept-new` automatically accepts GitHub's host key the first time we connect. (Without it, ssh would ask you to type `yes`, which a notebook cell cannot do.)

````bash
ssh -T -o StrictHostKeyChecking=accept-new git@github.com
````

If it worked, you will see a message like this (with your own username):

```
Hi your-github-username! You've successfully authenticated, but GitHub does not provide shell access.
```
That is a success. (The command may report an exit status of 1, which is also normal here.)

If you see `Permission denied (publickey)`, GitHub did not recognize your key. Go back and check that you copied the **whole** public key (and not the private key), and that you pasted it into **your** account.

### Is it really GitHub?
The first time you connect to a server, ssh remembers its host key, and it will warn you if it ever changes. To make sure that you connected to the real GitHub and not an impostor, compare the fingerprint that ssh saved with the one that GitHub publishes at [GitHub's SSH key fingerprints](https://docs.github.com/en/authentication/keeping-your-account-and-data-secure/githubs-ssh-key-fingerprints):

````bash
ssh-keygen -F github.com -l
````

# Get changes from GitHub with `git pull`
Pushing sends your commits **to** GitHub. `git pull` does the reverse: it downloads commits that are on GitHub but not on your computer yet. You need it when:

* you made changes from a different computer (say, your laptop),
* you edited a file in the GitHub web interface, or
* a collaborator pushed new commits to a repository that you share.

Let's pretend to be a collaborator by changing a file **on GitHub**:

1. In your browser, open your repository on GitHub, and click on `README.md`.
2. Click the **pencil icon** (Edit this file).
3. Add a new line of text, such as `Edited on GitHub!`.
4. Click **Commit changes...**, and then **Commit changes** again in the box that opens.

Your local copy does not know about this change yet:

````bash
cat README.md
````

Now pull it down:

````bash
git pull
````

````bash
cat README.md
````

````bash
git log --oneline --decorate
````

The commit that was made on GitHub is now in your local history.

**Tip:** make `git pull` the **first** thing you do whenever you sit down to work on a shared repository, and `git push` the last. That keeps you from editing an out-of-date copy.

# Troubleshooting
| Message | What it means | What to do |
|---|---|---|
| `fatal: not a git repository` | You are not inside a folder that has a `.git` folder. | Run `pwd` and `cd` to your repository (or `git init` if you meant to start one). |
| `Author identity unknown` / `Please tell me who you are` | You did not configure your name and email. | Run the two `git config --global` commands from the beginning of this guide. |
| `nothing to commit, working tree clean` | There is nothing new to save. | Check that you saved your file, and that you ran `git add` on it. |
| `nothing added to commit but untracked files present` | You ran `git commit` without staging the new files. | `git add` the files, and commit again. |
| `Permission denied (publickey)` | GitHub does not recognize your SSH key. | Repeat Steps 1-4 of the SSH section. Make sure the key you added is the **public** key from Kamiak, and that you used the SSH URL (`git@github.com:...`), not the HTTPS URL. |
| `src refspec main does not match any` | There is no branch named `main`, usually because you have not made a commit yet, or your branch is named `master`. | Make your first commit, and run `git branch -M main`. |
| `remote origin already exists` | You already added a remote called `origin`. | Look at it with `git remote -v`. Fix it with `git remote set-url origin NEW_URL`. |
| `Repository not found` | The URL has a typo (often the username), or the repository is private and you are not logged in as its owner. | Compare the URL from `git remote -v` with the one on GitHub, and fix it with `git remote set-url`. |
| `rejected ... (fetch first)` or `Updates were rejected because the remote contains work that you do not have locally` | There are commits on GitHub that you do not have yet. | Run `git pull`, then `git push` again. |
| `CONFLICT (content): Merge conflict in file` | You and GitHub both changed the same lines of a file. | Open the file, and look for the lines between `<<<<<<<` and `>>>>>>>`. Edit the file to keep what you want and delete the marker lines, then `git add` the file and `git commit`. If you are stuck, ask for help before you try anything else! |
| `error: Your local changes would be overwritten by merge` | `git pull` would overwrite changes that you have not committed. | Commit your changes first (or `git restore` the file if you do not want them), then pull again. |

When something goes wrong, `git status` is always a good first step. It tells you what git thinks is going on, and usually what to do next.

# Cheat sheet
| Command | What it does |
|---|---|
| `git init` | Start tracking the current folder |
| `git status` | Show what is modified, staged, and untracked |
| `git add FILE` | Stage a file (`git add .` stages everything) |
| `git commit -m "message"` | Save the staged changes as a commit |
| `git diff` | Show changes that are not staged yet |
| `git log --oneline` | Show the commit history |
| `git restore FILE` | Discard uncommitted changes to a file |
| `git restore --staged FILE` | Unstage a file |
| `git restore --source=HASH FILE` | Bring back an older version of a file |
| `git remote add origin URL` | Connect your repo to a repo on GitHub |
| `git remote -v` | Show your remotes |
| `git push` | Send your commits to GitHub |
| `git pull` | Get new commits from GitHub |
| `git clone URL` | Download a copy of a repository |

**The routine:** `git pull` &rarr; edit your files &rarr; `git status` &rarr; `git add` &rarr; `git commit -m "..."` &rarr; `git push`
